in this notebook we will demostate to do the following : 
- POS tagging
- Dependency parsing.

the first step is setup where we import nltk and download the dependancies 

```python
import nltk

nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
nltk.download("averaged_perceptron_tagger", quiet=True)
nltk.download("averaged_perceptron_tagger_eng", quiet=True)
```

first we will demostate how to do POS taging 

```python
sentence = "The quick fox jumps over the lazy dog"

tokens = nltk.word_tokenize(sentence)
pos_tags = nltk.pos_tag(tokens)

print(pos_tags)
```

the output can be interpretend as follows : 
Tag meanings: DT = determiner, JJ = adjective, NN = noun, VBZ = verb (3rd person singular), IN = preposition.

next we will demostate dependancy parsing 

```python
from nltk.parse import DependencyGraph
from nltk.parse.dependencygraph import DependencyGraph

# Grammar format: 'head' -> 'dependent1' | 'dependent2'
grammar = nltk.DependencyGrammar.fromstring("""
    'jumps' -> 'fox' | 'over'
    'fox' -> 'The' | 'quick'
    'over' -> 'dog'
    'dog' -> 'the' | 'lazy'
""")

parser = nltk.ProjectiveDependencyParser(grammar)

for tree in parser.parse(tokens):
    print(tree)
    tree.pretty_print()
```

How to read it:

jumps is the root (main verb).
fox is the subject of jumps; The and quick depend on fox.
over is attached to jumps, and dog depends on over; the and lazy depend on dog.
