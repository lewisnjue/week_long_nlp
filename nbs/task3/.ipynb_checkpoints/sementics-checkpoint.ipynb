```python
def get_sentiment(text):
    polarity = TextBlob(text).sentiment.polarity
    if polarity > 0:
        label = "Positive"
    elif polarity < 0:
        label = "Negative"
    else:
        label = "Neutral"
    return polarity, label

results = []
for tweet in tweets:
    polarity, label = get_sentiment(tweet)
    results.append({"Tweet": tweet, "Polarity": round(polarity, 2), "Sentiment": label})

df = pd.DataFrame(results)
df
```

The overall score is the average polarity across all tweets.

```python
overall_score = df["Polarity"].mean()

if overall_score > 0:
    overall_label = "Positive"
elif overall_score < 0:
    overall_label = "Negative"
else:
    overall_label = "Neutral"

print(f"Overall sentiment score: {overall_score:.2f}")
print(f"Overall sentiment: {overall_label}")
```
