in this notebook we will demostrate how to do preprocessing using python and nltk library 

```python
import nltk
```

```python
# sample text : 
text = "Just finished running 5 miles and I'm feeling amazing! #fitness"
```

the first step will be tokenization, so for that we will do the following : 

```python
nltk.download('punkt')
nltk.download('punkt_tab') # more resent
from nltk.tokenize import word_tokenize, sent_tokenize

```

for tokenization we can use word tokenization or we can use sentence tokenization we will show them both here 

```python
words = word_tokenize(text)
sent = sent_tokenize(text)
print("Word tokenization", words)
print("Sentence tokenization", sent)
```

The other preprocessing we are going to look at is stop word removal and we can do the following to achieve that , 

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

filtered = [ w for w in words if w.lower() not in stop_words]
print(filtered)
```

Then we can perform stemming and for this we will use poster stemmer for simplicity 

```python

from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
stems  =   [ stemmer.stem(word) for word in words]
print(stems)
```

we can instead of using stemming use lammetization we can be done as it follows 

```python

nltk.download('wordnet')
nltk.download('omw-1.4')
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmas = [ lemmatizer.lemmatize(word ) for word in words ]
print(lemmas)

```
